# CHỦ ĐỀ 1: TỐI ƯU HÓA HÀNH TRÌNH KHÁCH HÀNG & HIỆU QUẢ MARKETING

## 1. Bối cảnh & Đặt vấn đề tổng thể
Trong ngành thương mại điện tử, chi phí để mang một lượt truy cập (traffic) vào website và chuyển đổi họ thành khách hàng là rất đắt đỏ. Khách hàng hiếm khi thực hiện giao dịch ngay lần đầu chạm mặt thương hiệu; thay vào đó, họ trải qua một hành trình phức tạp (Customer Journey) từ nhận thức, cân nhắc, chốt đơn, cho đến mua lặp lại.

Qua topic này nhà doanh nghiệp có thể:
- **Bóc tách được vai trò thực sự của các kênh Marketing:** Kênh nào thu hút khách mới, kênh nào chốt nhiều đơn?
- **Hiểu rõ nhịp độ chốt đơn:** Khách hàng cần bao nhiêu thời gian để ra quyết định? Nhóm nào đang đăng ký nhưng không mua?
- **Đánh giá đúng chất lượng khách hàng:** Đo lường hiệu quả quảng cáo đang bị "ảo" nếu chỉ nhìn vào số lượng đăng ký mà bỏ qua Giá trị vòng đời khách hàng (LTV).

## 2. Mục tiêu phân tích
Báo cáo này được thực hiện nhằm mục đích xem xét toàn bộ hành trình trên thông qua 3 chặng phân tích chính:
- **Chặng 1:** Phân tích Mô hình phân bổ Kênh (Attribution Model) - Truy tìm điểm chạm đầu và cuối.
- **Chặng 2:** Phân tích Hành vi & Độ trễ chuyển đổi (Time-to-Conversion).
- **Chặng 3:** Phân tích Chất lượng & Giá trị Vòng đời khách hàng (LTV & Retention) theo từng kênh.

Từ đó, đưa ra các đề xuất hành động (Prescriptive Actions) nhằm phân bổ lại ngân sách Marketing, tối ưu hóa điểm chạm và tăng trưởng doanh thu bền vững.

---
**Giải thích thuật ngữ** 

Giá trị vòng đời khách hàng (Customer Lifetime Value - LTV):
- Giải thích đơn giản: Tổng số tiền mà một khách hàng chi trả cho shop trong suốt cuộc đời của họ (từ lúc bắt đầu mua đến lúc họ bỏ shop).
- Công thức "coder": SUM(all_payments_by_user_id).
- Tại sao quan trọng? Nếu chi phí quảng cáo để có 1 user là 100k, mà LTV của họ chỉ có 50k là bạn đang "burn money" (đốt tiền).

Độ trễ chuyển đổi (Time-to-Conversion):
- Giải thích đơn giản: Khoảng thời gian từ lúc User thực hiện hành động A (ví dụ: Signup) đến khi thực hiện hành động B (ví dụ: Mua đơn đầu tiên).
- Công thức "coder": DATEDIFF(first_order_date, signup_date).
- Ý nghĩa: Giúp doanh nghiệp biết khi nào nên gửi thông báo "Push Notification" hoặc Email nhắc nhở để khách không quên giỏ hàng.

Tỷ lệ giữ chân (Retention):
- Giải thích đơn giản: Khả năng shop giữ được khách hàng cũ quay lại mua lần tiếp theo.
- Dưới góc độ hệ thống: Nếu Retention thấp, hệ thống của bạn giống như một "cái xô thủng", bạn cứ đổ thêm User mới vào thì User cũ lại chảy ra hết.

## **BƯỚC 1: PHÂN TÍCH MÔ HÌNH PHÂN BỔ KÊNH (First-touch vs. Last-touch)**
### 1. Bài toán
Xác định luồng dịch chuyển kênh tiếp thị của khách hàng: từ điểm chạm đầu tiên khiến họ đăng ký tài khoản (acquisition_channel - First-touch) cho đến điểm chạm cuối cùng dẫn đến việc họ "chốt" đơn hàng đầu tiên (order_source - Last-touch).

### 2. Động lực thực hiện
Trong mô hình thương mại điện tử, việc gán toàn bộ công trạng (và ngân sách marketing) cho kênh tạo ra đơn hàng (Last-touch) là một sai lầm chết người. Ví dụ: Nếu một khách hàng thấy quảng cáo trên Facebook, ấn vào xem rồi đăng ký tài khoản (nhưng chưa mua). Vài ngày sau, họ nhớ ra, lên Google gõ tên shop (Organic Search) và mua hàng. Nếu chỉ nhìn vào bảng Orders, ta sẽ nghĩ Facebook vô dụng và dồn tiền cho SEO/Google.
Ma trận chuyển dịch này sẽ minh oan cho các kênh "nuôi dưỡng" (Top-of-funnel) và chỉ rõ kênh nào thực sự làm nhiệm vụ "chốt sale" (Bottom-of-funnel).

Vị trí của bảng web_traffic.csv ở đâu?
Như bạn đã phát hiện, web_traffic.csv bị gộp theo ngày và nguồn mạnh nhất. Nên chúng ta sẽ tách nó ra thành một mục riêng biệt chuyên dùng để Khám phá Xu hướng Vĩ mô (Macro Trend & Seasonality). Ta sẽ vẽ biểu đồ đường (Line chart) xem lượng sessions và bounce rate thay đổi ra sao qua các năm/tháng, có trùng khớp với các mùa sale lớn hay không.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Đọc dữ liệu
df_cust = pd.read_csv('../dataset/customers.csv')
df_orders = pd.read_csv('../dataset/orders.csv')

# 2. Lấy đơn hàng ĐẦU TIÊN của mỗi khách hàng
# Sắp xếp theo ngày đặt hàng và group theo khách hàng để lấy dòng đầu tiên
first_orders = df_orders.sort_values('order_date').groupby('customer_id').first().reset_index()

# 3. Gộp bảng theo customer_id để đối chiếu Kênh đăng ký và Kênh mua hàng
merged_df = pd.merge(
    first_orders[['customer_id', 'order_source']], 
    df_cust[['customer_id', 'acquisition_channel']], 
    on='customer_id', 
    how='inner'
)

# 4. Tạo Ma trận phân bổ (Cross-tabulation)
cross_tab = pd.crosstab(
    merged_df['acquisition_channel'], 
    merged_df['order_source'], 
    margins=True, 
    margins_name="Total"
)

print("Ma trận chuyển dịch Kênh tiếp thị (Acquisition vs First Order Source):")
print(cross_tab)

# 5. Vẽ Heatmap (bỏ qua cột/hàng Total để biểu đồ không bị lệch scale)
plt.figure(figsize=(10, 6))
heatmap_data = pd.crosstab(merged_df['acquisition_channel'], merged_df['order_source'])
sns.heatmap(heatmap_data, annot=True, fmt='d', cmap='Blues')
plt.title('Hành trình chuyển đổi: Kênh Đăng ký (Y) vs Kênh Chốt Đơn Đầu Tiên (X)')
plt.ylabel('Kênh thu hút (Acquisition Channel)')
plt.xlabel('Kênh chốt đơn (Order Source)')
plt.tight_layout()
```

---
### Giải thích các Kênh Marketing (Acquisition Channels)

| Kênh Marketing | Phân loại | Giải thích hành vi khách hàng |
| :--- | :--- | :--- |
| **Organic Search** | Tự nhiên (Không trả tiền) | Khách gõ "váy dự tiệc" $\rightarrow$ Thấy shop ở trang 1 Google $\rightarrow$ Click |

| **Referral** | Giới thiệu | Được giới thiệu từ website, blog khác hoặc từ bạn bè qua link Affiliate |
| **Direct** | Trực tiếp | Khách gõ trực tiếp tên miền (abc.com) vào thanh địa chỉ trình duyệt |


### 3. Nhận xét Chẩn đoán (Diagnostic):

- Sự thật về Social Media (Mạng xã hội): Nhìn vào hàng "Social Media", kênh này mang về tới 18,002 khách hàng mới (Acquisition). Tuy nhiên, chỉ có khoảng 20% trong số đó (3,616 người) là chốt đơn trực tiếp qua link Social. Một lượng khổng lồ hơn 5,000 người đăng ký từ mạng xã hội, nhưng lại chốt đơn bằng cách tự tìm kiếm trên Google (Organic Search) hoặc click vào Quảng cáo tìm kiếm (Paid Search).
Insight: Social Media đóng vai trò gieo rắc nhận thức cực tốt, nhưng khách hàng thường không mua ngay khi lướt Feed mà sẽ thoát ra và tìm lại trên Google khi họ thực sự có nhu cầu.

- Sức mạnh tuyệt đối của Organic Search (Tìm kiếm tự nhiên):
Đây là kênh "Vua" ở cả hai đầu phễu. Nó vừa hút khách đăng ký nhiều nhất (26,950 người) vừa là kênh chốt đơn cuối cùng mạnh nhất (25,603 đơn). Điều này chứng tỏ thương hiệu shop đã có "Top of Mind" (độ nhận diện) cực kỳ cao trên thị trường, hoặc SEO của website đang làm quá xuất sắc.

### 4. Đề xuất hành động (Prescriptive):

- Không cắt giảm ngân sách Social Media: Dù ROAS (Return On Ad Spend) trực tiếp của các chiến dịch Facebook/Instagram/TikTok có vẻ thấp (số đơn chốt trực tiếp ít), tuyệt đối không được tắt đi. Vì nếu tắt, cột "Organic Search" và "Paid Search" cũng sẽ sụt giảm theo (do khách không còn được mồi nhận thức nữa).

- Chiến lược đồng bộ (Omni-channel): Vì có tới hơn 3,900 khách biết qua mạng xã hội rồi mới lên Google Search Ads để mua. Doanh nghiệp cần đồng bộ từ khóa: Các nội dung giật tít, tên bộ sưu tập chạy trên Social cần được mua luôn từ khóa trên Google Search Ads. Để khi khách lướt thấy hay, lên Google gõ tên bộ sưu tập đó thì quảng cáo trang web của mình phải hiện ra đầu tiên, tránh bị đối thủ cướp mất "miếng mồi" đã được nuôi dưỡng.

## **Bước 2: Phân tích Hành vi Chuyển đổi & Vòng đời Khách hàng**
### 1. Bài toán:

- Đo lường tỷ lệ khách hàng "ngủ đông" (Đăng ký nhưng không bao giờ mua).

- Xử lý bài toán thời gian thực: Tỷ lệ khách mua tức thì (Same Day), khách vãng lai (Guest/Before Signup) và khách cần nuôi dưỡng (Later).

- Đo lường tốc độ chốt đơn dựa trên nhân khẩu học (age_group, gender).

- Tính toán thời gian vòng đời (Khách mất bao lâu để quay lại mua lần 2, lần 3).

### 2. Động lực thực hiện:
Thay vì nhìn khách hàng như một khối đồng nhất, việc "bóc tách" này giúp cá nhân hóa toàn bộ thông điệp Marketing. Nếu biết độ tuổi 18-24 có độ trễ mua hàng khác với 45-54, ta có thể cài đặt kịch bản gửi Email Remind (Nhắc nhở) khác nhau. Việc biết "Thời gian quay lại" giúp tối ưu hóa điểm rơi của các mã giảm giá (Voucher) cho khách hàng cũ.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Đọc và chuẩn bị dữ liệu
df_cust = pd.read_csv('../dataset/customers.csv')
df_orders = pd.read_csv('../dataset/orders.csv')

df_cust['signup_date'] = pd.to_datetime(df_cust['signup_date'])
df_orders['order_date'] = pd.to_datetime(df_orders['order_date'])

# ==========================================
# BIỂU ĐỒ 1: Tỷ lệ "Ngủ đông" vs "Có mua hàng"
# ==========================================
buyers = df_orders['customer_id'].unique()
df_cust['has_purchased'] = df_cust['customer_id'].isin(buyers)

plt.figure(figsize=(8, 4))
sns.barplot(
    data=df_cust.groupby('acquisition_channel')['has_purchased'].mean().reset_index(),
    x='has_purchased', y='acquisition_channel', color='#3498db'
)
plt.title('Tỷ lệ khách hàng Đăng ký & Đã chốt đơn theo Kênh')
plt.xlabel('Tỷ lệ chuyển đổi (%)')
plt.ylabel('Kênh tiếp thị')
plt.show()

# ==========================================
# BIỂU ĐỒ 2: Hành vi "Mua trước - Đăng ký sau"
# ==========================================
first_orders = df_orders.sort_values('order_date').groupby('customer_id')['order_date'].first().reset_index()
first_orders.columns = ['customer_id', 'first_order_date']
df_merged = pd.merge(df_cust[df_cust['has_purchased']], first_orders, on='customer_id')

df_merged['days_to_convert'] = (df_merged['first_order_date'] - df_merged['signup_date']).dt.days
df_merged['convert_type'] = np.where(df_merged['days_to_convert'] < 0, 'Mua trước khi ĐK (Guest)',
                            np.where(df_merged['days_to_convert'] == 0, 'Mua ngay (Same Day)', 'Mua sau (Later)'))
 
plt.figure(figsize=(6, 4))
df_merged['convert_type'].value_counts().plot.pie(autopct='%1.1f%%', colors=['#2ecc71', '#e74c3c', '#f1c40f'])
plt.title('Phân loại hành vi Chốt đơn đầu tiên')
plt.ylabel('')
plt.show()

# ==========================================
# BIỂU ĐỒ 3: Độ trễ chuyển đổi theo Nhân khẩu học (Chỉ xét nhóm Later)
# ==========================================
later_buyers = df_merged[df_merged['days_to_convert'] > 0]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=later_buyers, x='age_group', y='days_to_convert', ax=axes[0], palette='Set2')
axes[0].set_title('Độ trễ mua hàng theo Độ tuổi')
axes[0].set_ylabel('Số ngày (Days)')

sns.boxplot(data=later_buyers, x='gender', y='days_to_convert', ax=axes[1], palette='Set3')
axes[1].set_title('Độ trễ mua hàng theo Giới tính')
axes[1].set_ylabel('')
plt.tight_layout()
plt.show()

# ==========================================
# BIỂU ĐỒ 4: Tần suất quay lại (Inter-order gap)
# ==========================================
df_orders_sorted = df_orders.sort_values(['customer_id', 'order_date'])
df_orders_sorted['prev_order_date'] = df_orders_sorted.groupby('customer_id')['order_date'].shift(1)
df_orders_sorted['inter_order_gap'] = (df_orders_sorted['order_date'] - df_orders_sorted['prev_order_date']).dt.days

# ==========================================
# BIỂU ĐỒ 4: Inter-order gap (BỔ SUNG PHÂN LOẠI)
# ==========================================
df_orders_sorted['order_seq'] = df_orders_sorted.groupby('customer_id').cumcount() + 1

# Thêm annotation về ý nghĩa
print(f"Median inter-order gap (tất cả các cặp đơn liên tiếp): {df_orders_sorted['inter_order_gap'].median():.0f} ngày")
print(f"Median gap từ đơn 1 -> 2: {df_orders_sorted[df_orders_sorted['order_seq']==2]['inter_order_gap'].median():.0f} ngày")

plt.figure(figsize=(8, 4))
sns.histplot(df_orders_sorted['inter_order_gap'].dropna(), bins=50, kde=True, color='purple')
plt.title('Phân phối số ngày giữa 2 lần mua liên tiếp (Retention)')
plt.xlabel('Số ngày giữa các đơn (Days)')
plt.ylabel('Số lượng đơn hàng')
plt.xlim(0, 700) # Giới hạn trục X để tránh nhiễu ngoại lai
plt.show()
```

## **BƯỚC 3: PHÂN TÍCH CHẤT LƯỢNG KHÁCH HÀNG (RETENTION & LTV)**
### 1. Mục tiêu cốt lõi
Trả lời câu hỏi chiến lược: "Kênh nào mang lại những khách hàng 'sộp' nhất và trung thành nhất?" Thay vì đếm số lượng người đăng ký, chúng ta sẽ đo lường "chất lượng" của họ thông qua hai hệ số:

- Frequency (Tần suất): Trung bình một khách hàng từ kênh này mua bao nhiêu đơn? (Đại diện cho lòng trung thành / Retention).

- Monetary (Giá trị tiền tệ): Trung bình một khách hàng từ kênh này mang lại bao nhiêu tổng doanh thu cho công ty? (Đại diện cho LTV).

### 2. Ý nghĩa kinh doanh (Business Logic) & Tránh bẫy "CPA rẻ"
Trong thực tế chạy quảng cáo, các marketer rất hay rơi vào bẫy "CPA rẻ" (Cost Per Acquisition - Chi phí trên mỗi lượt đăng ký). Ví dụ:

Chạy Facebook Ads (Social Media) tốn 10,000đ/đăng ký. Khách vào mua 1 cái áo phông rồi không bao giờ quay lại.

Chạy Google Search (Paid Search) tốn 50,000đ/đăng ký. Nhìn có vẻ đắt gấp 5 lần, nhưng khách này quay lại mua 10 lần trong 2 năm tiếp theo.
Nếu chỉ nhìn bề mặt, ta sẽ dồn tiền cho Social Media. Nhưng phân tích LTV ở Bước 3 sẽ lật ngược lại vấn đề, giúp Ban Giám Đốc đưa ra quyết định đầu tư khôn ngoan trong dài hạn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Đọc và chuẩn bị dữ liệu
df_cust = pd.read_csv('../dataset/customers.csv')
df_orders = pd.read_csv('../dataset/orders.csv')
df_payments = pd.read_csv('../dataset/payments.csv')

# Bỏ qua đơn hàng đã hủy
valid_orders = df_orders[df_orders['order_status'] != 'cancelled']

# 2. Tính toán LTV và Frequency
order_values = df_payments.groupby('order_id')['payment_value'].sum().reset_index()
orders_with_value = pd.merge(valid_orders, order_values, on='order_id', how='inner')

customer_ltv = orders_with_value.groupby('customer_id').agg(
    Frequency=('order_id', 'nunique'),
    Monetary=('payment_value', 'sum')
).reset_index()

customer_full = pd.merge(customer_ltv, df_cust[['customer_id', 'acquisition_channel']], on='customer_id')

channel_ltv = customer_full.groupby('acquisition_channel').agg(
    Avg_Frequency=('Frequency', 'mean'),
    Avg_LTV=('Monetary', 'mean'),
    Customer_Count=('customer_id', 'count')
).reset_index()

# Thiết lập phong cách tối giản
sns.set_theme(style="whitegrid")

# ==========================================
# BIỂU ĐỒ 1: Kênh nào mang về NHIỀU KHÁCH NHẤT? (Volume)
# ==========================================
plt.figure(figsize=(8, 4))
data_count = channel_ltv.sort_values('Customer_Count', ascending=False)
sns.barplot(data=data_count, x='Customer_Count', y='acquisition_channel', palette='Blues_r')
plt.title('1. Quy mô tập khách hàng theo Kênh', fontweight='bold')
plt.xlabel('Số lượng khách hàng')
plt.ylabel('')
plt.show()

# ==========================================
# BIỂU ĐỒ 2: Kênh nào có KHÁCH MUA LẶP LẠI NHIỀU NHẤT? (Retention/Frequency)
# ==========================================
plt.figure(figsize=(8, 4))
# Cắt trục X bắt đầu từ 6.5 để làm nổi bật sự chênh lệch (vì các con số rất sát nhau)
data_freq = channel_ltv.sort_values('Avg_Frequency', ascending=False)
sns.barplot(data=data_freq, x='Avg_Frequency', y='acquisition_channel', palette='Greens_r')
plt.title('2. Tần suất mua hàng trung bình (Số đơn/Khách)', fontweight='bold')
plt.xlabel('Số đơn hàng')
plt.ylabel('')
plt.xlim(6.5, 6.75) 
plt.show()

# ==========================================
# BIỂU ĐỒ 3: Kênh nào mang lại DOANH THU CAO NHẤT trên mỗi khách? (LTV)
# ==========================================
plt.figure(figsize=(8, 4))
data_ltv = channel_ltv.sort_values('Avg_LTV', ascending=False)
sns.barplot(data=data_ltv, x='Avg_LTV', y='acquisition_channel', palette='Oranges_r')
plt.title('3. Giá trị Vòng đời Khách hàng (LTV)', fontweight='bold')
plt.xlabel('Tổng chi tiêu trung bình (VNĐ)')
plt.ylabel('')
plt.xlim(150000, 165000) # Cắt trục X để dễ so sánh
plt.show()
```

### 3. Phân tích và Nhận xét (Góc nhìn mới, sắc nét hơn)

Nhờ việc tách biệt 3 biểu đồ ngang, chúng ta thấy rõ một nghịch lý rất thú vị của bộ dữ liệu này: "Kênh mang về nhiều khách nhất chưa chắc đã là kênh khách chi tiêu xộp nhất".

Khám phá từ Biểu đồ Quy mô (Volume):

- Organic Search (Tìm kiếm tự nhiên) là "con gà đẻ trứng vàng" về mặt số lượng khi mang về hơn 26,000 khách hàng, áp đảo hoàn toàn các kênh còn lại (Paid Search và Social Media chỉ quanh mức 17,500).

Khám phá từ Biểu đồ Tần suất (Retention):

- Nhìn chung, khả năng giữ chân khách hàng của doanh nghiệp cực kỳ đồng đều, trung bình đạt ~6.6 đến 6.7 đơn/người.

- Tuy nhiên, khách tự tìm kiếm (Organic Search) vẫn dẫn đầu về mức độ trung thành (6.7 đơn/người). Trong khi đó, nhóm khách được giới thiệu (Referral) lại chót bảng. Điều này chứng tỏ khách mua vì nhu cầu thật luôn quay lại nhiều hơn khách mua vì nể bạn bè giới thiệu.

Khám phá từ Biểu đồ LTV (Cú "Twist" bất ngờ):

- Mặc dù Organic Search đông nhất và mua nhiều lần nhất, nhưng Social Media mới là kênh mang lại LTV cao nhất (hơn 163,000).

- Nghĩa là: Khách hàng đến từ Facebook/Instagram có thể mua ít đơn hơn một chút, nhưng mỗi lần mua họ lại chốt những món đồ giá trị cao hơn (Giá trị trung bình trên một đơn hàng - AOV của họ cao hơn).

### 4. Đề xuất chiến lược (Prescriptive):

- Đối với Social Media (Chuyên trị bán hàng giá cao): Đừng dùng mạng xã hội để chạy quảng cáo các sản phẩm xả kho hay giá rẻ. Hãy đẩy mạnh quảng bá các bộ sưu tập giới hạn (Limited Edition), các mặt hàng Premium trên Social Media vì tập khách hàng ở đây rất chịu chi.

- Đối với Organic Search (Chuyên trị bán số lượng lớn): Tối ưu hóa SEO cho các mặt hàng cơ bản (Basic wear, Standard segment) vì nhóm khách này có xu hướng mua lặp lại các mặt hàng thiết yếu rất thường xuyên.

- Kiểm tra lại tệp Referral: Khách từ kênh Giới thiệu đang mua ít nhất và chi tiêu cũng thấp nhất (áp chót). Cần thay đổi cơ chế giới thiệu: chỉ thưởng tiền khi người được giới thiệu mua đơn hàng trên 500k, để lọc bớt những khách hàng "săn sale" giá trị thấp.
